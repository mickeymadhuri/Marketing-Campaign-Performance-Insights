Campaign Performance Analysis, Scatter Plot Acquisition_Cost vs ROI

```python
plt.figure(figsize=(10, 6))

sns.scatterplot(
    data=df,
    x='Acquisition_Cost',
    y='ROI',
    hue='Campaign_Type'
)

plt.title('Acquisition Cost vs ROI')
plt.xlabel('Acquisition Cost')
plt.ylabel('ROI')
plt.legend(title='Campaign Type')
plt.show()
```

Bar Chart Average Conversion Rate by Channel_Used, categorized by Campaign_Type

```python
plt.figure(figsize=(12, 6))

sns.barplot(
    data=df,
    x='Channel_Used',
    y='Conversion_Rate',
    hue='Campaign_Type',
    errorbar=None
)

plt.title('Average Conversion Rate by Channel and Campaign Type')
plt.xlabel('Channel Used')
plt.ylabel('Average Conversion Rate')
plt.xticks(rotation=45)
plt.legend(title='Campaign Type')
plt.tight_layout()
plt.show()
```

Box Plot – Engagement_Score by Campaign_Type

```python
plt.figure(figsize=(10, 6))

sns.boxplot(
    data=df,
    x='Campaign_Type',
    y='Engagement_Score'
)

plt.title('Engagement Score Distribution by Campaign Type')
plt.xlabel('Campaign Type')
plt.ylabel('Engagement Score')
plt.xticks(rotation=45)
plt.show()
```

Bar Chart – Average ROI by Company

```python
company_roi = df.groupby('Company')['ROI'].mean().reset_index()

plt.figure(figsize=(12, 6))

sns.barplot(
    data=company_roi,
    x='Company',
    y='ROI',
    errorbar=None
)

plt.title('Average ROI by Company')
plt.xlabel('Company')
plt.ylabel('Average ROI')
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
```

 Heatmap – Engagement_Score and Conversion_Rate

```python
plt.figure(figsize=(8, 5))

correlation = df[['Engagement_Score', 'Conversion_Rate']].corr()

sns.heatmap(
    correlation,
    annot=True,
    cmap='coolwarm',
    fmt='.2f'
)

plt.title('Correlation Between Engagement Score and Conversion Rate')
plt.show()
```

Customer Segmentation, Count Plot – Distribution of Target_Audience

```python
plt.figure(figsize=(12, 6))

sns.countplot(
    data=df,
    x='Target_Audience',
    order=df['Target_Audience'].value_counts().index
)

plt.title('Distribution of Target Audience')
plt.xlabel('Target Audience')
plt.ylabel('Number of Campaigns')
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
```

Bar Chart – Highest Conversion_Rate Customer_Segment for Each Language

```python
segment_language = (
    df.groupby(['Language', 'Customer_Segment'])['Conversion_Rate']
    .mean()
    .reset_index()
)

plt.figure(figsize=(12, 6))

sns.barplot(
    data=segment_language,
    x='Language',
    y='Conversion_Rate',
    hue='Customer_Segment',
    errorbar=None
)

plt.title('Conversion Rate by Customer Segment and Language')
plt.xlabel('Language')
plt.ylabel('Average Conversion Rate')
plt.legend(title='Customer Segment', bbox_to_anchor=(1.05, 1))
plt.tight_layout()
plt.show()
```

```python
highest_segment = segment_language.loc[
    segment_language.groupby('Language')['Conversion_Rate'].idxmax()
]

print(highest_segment)
```

Box Plot – Acquisition_Cost by Customer_Segment and Channel_Used

Histogram

Line chart

```python
date_conversion = (
    df.groupby(['Date', 'Company'])['Conversion_Rate']
    .mean()
    .reset_index()
)

plt.figure(figsize=(14, 7))

sns.lineplot(
    data=date_conversion,
    x='Date',
    y='Conversion_Rate',
    hue='Company',
    marker='o'
)

plt.title('Conversion Rate Trends Over Time by Company')
plt.xlabel('Date')
plt.ylabel('Average Conversion Rate')
plt.xticks(rotation=45)
plt.legend(title='Company', bbox_to_anchor=(1.05, 1))
plt.tight_layout()
plt.show()
```

```python
date_engagement = (
    df.groupby('Date')['Engagement_Score']
    .mean()
    .reset_index()
)

plt.figure(figsize=(12, 6))

sns.lineplot(
    data=date_engagement,
    x='Date',
    y='Engagement_Score',
    marker='o'
)

plt.title('Engagement Score Trend Over Time')
plt.xlabel('Date')
plt.ylabel('Average Engagement Score')
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
```

```python
location_cost = (
    df.groupby('Location')['Acquisition_Cost']
    .mean()
    .reset_index()
    .sort_values(by='Acquisition_Cost', ascending=False)
)

plt.figure(figsize=(12, 6))

sns.barplot(
    data=location_cost,
    x='Location',
    y='Acquisition_Cost',
    errorbar=None
)

plt.title('Average Acquisition Cost by Location')
plt.xlabel('Location')
plt.ylabel('Average Acquisition Cost')
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()

print("Location with Highest Acquisition Cost:")
print(location_cost.iloc[0])
```

```python
plt.figure(figsize=(14, 7))

sns.barplot(
    data=df,
    x='Location',
    y='Conversion_Rate',
    hue='Target_Audience',
    errorbar=None
)

plt.title('Conversion Rate by Location and Target Audience')
plt.xlabel('Location')
plt.ylabel('Average Conversion Rate')
plt.xticks(rotation=45)
plt.legend(title='Target Audience', bbox_to_anchor=(1.05, 1))
plt.tight_layout()
plt.show()
```

```python
location_roi = df.groupby('Location')['ROI'].sum()

plt.figure(figsize=(9, 9))

plt.pie(
    location_roi,
    labels=location_roi.index,
    autopct='%1.1f%%',
    startangle=90
)

plt.title('Percentage Distribution of ROI by Location')
plt.show()
```
